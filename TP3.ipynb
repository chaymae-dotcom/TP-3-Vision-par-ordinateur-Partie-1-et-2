Lecture et affichage d’une image : 

```python
import cv2

# 1. Charger l'image originale
img = cv2.imread('images.jpg')

# 2. Afficher l'image dans une fenêtre OpenCV
cv2.imshow('Image originale', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

Conversion en niveaux de gris : 

```python
import cv2

# Conversion de l'image originale en niveaux de gris
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Affichage 
cv2.imshow('Image en niveaux de gris', gray)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

Détection des contours : 

```python
import cv2

# Détection des contours avec l'algorithme de Canny
edges = cv2.Canny(img, 100, 200)

# Affichage 
cv2.imshow('Contours détectés', edges)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

<p style="margin-top: 12px;">
  <span style="background-color: #e0e7ff; color: #1d4ed8; font-weight: bold; padding: 2px 8px; border-radius: 4px; margin-right: 6px;">
    Question 1
  </span>
  Que représente la matrice retournée par OpenCV pour l’image ? 
</p>

Pour une image en couleur : C'est une matrice à 3 dimensions de taille $(hauteur \times largeur \times 3)$. Chaque pixel est représenté par un vecteur de 3 valeurs correspondant aux canaux de couleur BGR (Bleu, Vert, Rouge), comprises entre 0 et 255.

Pour une image en niveaux de gris : C'est une matrice à 2 dimensions de taille $(hauteur \times largeur)$. Chaque case (pixel) contient une seule valeur numérique représentant son intensité lumineuse, allant de 0 (noir total) à 255 (blanc pur).

<p style="margin-top: 12px;">
  <span style="background-color: #e0e7ff; color: #1d4ed8; font-weight: bold; padding: 2px 8px; border-radius: 4px; margin-right: 6px;">
    Question 2
  </span>
  Pourquoi convertit-on souvent l’image en niveaux de gris avant traitement ? 
</p>


Réduction de la complexité de calcul : On passe de 3 canaux de couleurs (BGR) à un seul canal, ce qui réduit considérablement le volume de données à traiter et accélère l'exécution des algorithmes.

Simplification de l'analyse : De nombreux algorithmes de vision par ordinateur (comme la détection de contours) s'intéressent uniquement aux variations d'intensité lumineuse et n'ont pas besoin de l'information de couleur, ce qui évite d'alourdir ou de biaiser les traitements

<p style="margin-top: 12px;">
  <span style="background-color: #e0e7ff; color: #1d4ed8; font-weight: bold; padding: 2px 8px; border-radius: 4px; margin-right: 6px;">

  Si on veut passer l’image en niveau de gris, avec un flou gaussien  
 par Exemple : blur = cv2.GaussianBlur(gray, (5, 5), 0) 
Que peut nous apporter le filtre Gaussien ? 
</p>


Réduction du bruit : Il permet de lisser l'image en atténuant les petites variations de pixels indésirables (le "bruit" numérique).

Amélioration des traitements futurs : C'est un pré-traitement indispensable (notamment avant l'algorithme de Canny) car il évite que le bruit de l'image ne soit interprété à tort comme de vrais contours, garantissant ainsi des résultats beaucoup plus nets et précis.

   Lecture vidéo / webcam 
</h3>

```python
import cv2

# Lecture d'une vidéo ou webcam
cap = cv2.VideoCapture(0)  

while True:
    ret, frame = cap.read()
    if not ret:
        break
    
    cv2.imshow('Webcam', frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
import cv2

cap = cv2.VideoCapture("BABYMONSTER - 'DESERVE' M V.mp4")

w = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
h = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter('sortie_gris.mp4', fourcc, 20.0, (w, h), isColor=False)

while True:
    ret, frame = cap.read()
    if not ret:  
        break
        
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    out.write(gray)
    
    cv2.imshow('Enregistrement', gray)
    if cv2.waitKey(25) & 0xFF == ord('q'): 
        break

cap.release()
out.release()
cv2.destroyAllWindows()
```

  Que se passe-t-il si vous utilisez une vidéo enregistrée à la place de la webcam ?

Lecture en accéléré : La vidéo se lit à une vitesse fulgurante . C'est parce que la boucle tourne à pleine puissance et que cv2.waitKey(1) ne marque qu'une pause minuscule d'une milliseconde, contrairement à la webcam qui impose son propre rythme de capture en temps réel.

Arrêt automatique : Contrairement à une webcam qui diffuse en continu indéfiniment, une vidéo enregistrée possède une fin. Une fois que toutes les images ont été lues, cap.read() renvoie False , ce qui déclenche le break et arrête proprement le programme.
